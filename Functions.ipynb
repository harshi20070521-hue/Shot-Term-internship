```python
#function without  any parameters
def sample():
  print("Good morning")
sample()
```

```python
#function with parameters
def sample(name):       # decleration
  print("My name is:",name)  #code
sample("Devayyani")# function call
```

```python
#function with return type
def add(a,b):
  return a+b # return call is used
print(add(2,3))
```

```python
# local variables used 
def f1():
    x = 30
    print("x value from f1:", x)
def f2():
    x = 34
    print("x value from f2:", x)
f1()
f2()
```

```python
# global variables are used
x = 10
def f1():
    print("x value from f1:", x)
def f2():
    print("x value from f2:", x)
f1()
f2()
print("x outside the function:", x)
```

```python
#arguments
def add(a,b):
  print(a+b)
add(20,30) # here it takes this number to run the function 
```

```python
#position arugument
def sample(a,b):
  print(a,b)
sample(2,3) #the a,b takes the values takes these first
sample(3,2) #now a,b takes these values
```

```python
# function
def name(first_name,last_name):
  print(first_name,last_name)
name("Rivan","Thakur")
```

```python
#keyword arugument
def sample(a,b):
  print(a,b)
sample(b=2,a=3)
```

```python
def greeting(msg, name):
    print("Hello", msg, name)
greeting(name="srivalli", msg="Good Morning")
```

```python
#default arugument
def student(name, dept="EEE"):
    print("My name:", name, "\n Department:", dept)
student(name="Rudra")
student(name="Rudra", dept="EEE")
```

```python
#variable arugument
def number(*n):
    print(n)
number(1)
number(2)
number(1, 2, 3)
```
