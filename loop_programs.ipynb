```python
# Print List Elements Using Index
l = [10, 20, 30, 40, 50]
print(l)
for i in range(0, len(l), 1):
    print(i)
    print(l[i], end=" ")
```

```python
# Multiplication Table 
num = int(input())
for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

```python
# Print Stars (One Star Per Line)
for i in range(5):
    for j in range(5):
        print("*")
    print()
```

```python
#Multiplication Tables from 1 to 10
for i in range(1, 11):
    print("Table of")
    for j in range(1, 11):
        print(j, "x", i, "=", j * i, end="\t")
    print()
```

```python
 # Square Star Pattern
for i in range(5):
    for j in range(5):
        print("*", end=" ")
    print()
```

```python
# Print Row Number (0–4)
for i in range(5):
    for j in range(5):
        print(i, end=" ")
    print()
```

```python
# Print Row Number (1–5)
for i in range(1, 6):
    for j in range(5):
        print(i, end=" ")
    print()
```

```python
# While Loop
i = 1
while i <= 10:
    print("Hello world")
    i += 1
```

```python
# Break Statement
for i in range(10):
    if i == 2:
        break
    print(i)

print("loop ended")
```

```python
# Break Using String Index
s = "Rivan"
for i in range(len(s)):
    if s[i] == 'a':
        break
    print(s[i])
print("loop ended")
```

```python
#Break Using Character
s = "Rivan"
for i in s:
    if i == 'n':
        break
    print(i)
print("loop ended")
```

```python
# Break Using User Input (Incorrect Comparison)
name = input("Enter your name: ")
var = input("var to break: ")
for i in name:
    if i == "var":
        break
    print(i)
print("loop ended")
```

```python
# Break Using User Input (Correct)
name = input("Enter your name: ")
var = input("var to break: ")
for i in name:
    if i == var:
        break
    print(i, end=" ")
print("loop ended")
```

```python
#Continue Statement
name = input("Enter your name: ")
var = input("var to break: ")
for i in name:
    if i == var:
        continue
    print(i)
print("loop ended")
```

```python
#Pass Statement
name = input("Enter your name: ")
var = input("var to break: ")
for i in name:
    if i == var:
        pass
    print(i, end=" ")
print("loop ended")
```

```python
# Descending Number Pattern
n = int(input("Enter n: "))
for i in range(n, 0, -1):
    for j in range(n):
        print(i, end=" ")
    print()
```

```python
# Reverse Column Pattern
n = int(input("Enter n: "))
for i in range(n):
    for j in range(n, 0, -1):
        print(j, end=" ")
    print()
```

```python
# Sequential Number Pattern
n = int(input("Enter n: "))
k = 1
for i in range(n):
    for j in range(n):
        print(k, end=" ")
        k += 1
    print()
```

```python
# Plus Pattern Using *
n = int(input("Enter n: "))
for i in range(n):
    for j in range(n):
        if j == 2 or i == 2:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
#Plus Pattern Using
n = int(input("Enter n: "))
for i in range(n):
    for j in range(n):
        if j == 2 or i == 2:
            print("#", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# Center Star Pattern
n = int(input("Enter n: "))
for i in range(n):
    for j in range(n):
        if j == 2 and i == 2:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# Alternate 1 and 0 Pattern
n = int(input("Enter n: "))
for i in range(n):
    for j in range(n):
        if j % 2 != 0:
            print(0, end=" ")
        else:
            print(1, end=" ")
    print()
```
