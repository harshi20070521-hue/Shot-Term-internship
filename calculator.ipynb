```python
import ipywidgets as widgets
from IPython.display import display, clear_output

a = widgets.FloatText(description='a', value=0)
b = widgets.FloatText(description='b', value=0)
op = widgets.Dropdown(options=['+','-','*','/'], description='op')
btn = widgets.Button(description='Compute')
out = widgets.Output()

def on_click(_):
    with out:
        clear_output()
        try:
            if op.value == '+':
                print(a.value + b.value)
            elif op.value == '-':
                print(a.value - b.value)
            elif op.value == '*':
                print(a.value * b.value)
            elif op.value == '/':
                if b.value == 0:
                    print('Error: division by zero')
                else:
                    print(a.value / b.value)
        except Exception as e:
            print('Error:', e)

btn.on_click(on_click)
display(widgets.HBox([a, b, op, btn]), out)
```
